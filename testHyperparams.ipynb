```python
from typing import Callable
import json
with open('hyperparams.json') as f:
    HYPERPARAMS = json.load(f)

# decide if I have to run Optuna
def isRecorded(questionnaire:str, model:str, scoringMetric:str) -> bool:
    global HYPERPARAMS
    if questionnaire in HYPERPARAMS:
        if model in HYPERPARAMS[questionnaire]:
            if scoringMetric in HYPERPARAMS[questionnaire][model]:
                return True
    return False

# decorator
def loadCached(model:str):

    def loadCached(optimizeFunction:Callable):
        global HYPERPARAMS

        def inner(questionnaire:str, scoringMetric:str):
            if isRecorded(questionnaire, model, scoringMetric):
                return HYPERPARAMS[questionnaire][model][scoringMetric]
            
            bestParams = optimizeFunction(questionnaire, scoringMetric)
            HYPERPARAMS[questionnaire][model][scoringMetric] = bestParams
            with open('hyperparams.json', 'w') as f:
                json.dump(HYPERPARAMS, f)
                
        return inner

    return loadCached
```
